# 7. Data Cleaning & Preprocessing

```python
# Create a copy of the original dataset
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
# Handle missing numerical values using median imputation

missing_columns = [
    "Rainfall_mm",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

for column in missing_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )

print("Missing values after imputation:")
print(df_clean[missing_columns].isnull().sum())

print("\nTotal missing values after imputation:",
      df_clean.isnull().sum().sum())
```

### Missing Value Treatment

The dataset contained 120 missing values in three numerical variables: rainfall, soil moisture, and yield.

Since these variables are continuous numerical measurements, the missing values were replaced using their respective median values. Median imputation was selected because it is less affected by extreme values than mean imputation.

After treatment, the dataset contains no missing values and is ready for further analysis.

```python
# Check duplicate rows

duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Check data types after preprocessing

df_clean.dtypes
```

```python
# Identify categorical and numerical columns

categorical_columns = df_clean.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numerical_columns = df_clean.select_dtypes(
    include=np.number
).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)

print("\nNumerical Columns:")
print(numerical_columns)
```

```python
# Check important categorical variables

for column in [
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]:
    print(f"\n{column}")
    print(df_clean[column].value_counts())
```

```python
# Check for logically invalid numerical values

validity_checks = {
    "Farm_Area_Hectares": (
        df_clean["Farm_Area_Hectares"] <= 0
    ).sum(),

    "Rainfall_mm": (
        df_clean["Rainfall_mm"] < 0
    ).sum(),

    "Humidity_pct": (
        (df_clean["Humidity_pct"] < 0) |
        (df_clean["Humidity_pct"] > 100)
    ).sum(),

    "Soil_Moisture_pct": (
        (df_clean["Soil_Moisture_pct"] < 0) |
        (df_clean["Soil_Moisture_pct"] > 100)
    ).sum(),

    "Yield_Tonnes_Ha": (
        df_clean["Yield_Tonnes_Ha"] < 0
    ).sum(),

    "Production_Tonnes": (
        df_clean["Production_Tonnes"] < 0
    ).sum(),

    "Total_Cost_INR": (
        df_clean["Total_Cost_INR"] < 0
    ).sum(),

    "Revenue_INR": (
        df_clean["Revenue_INR"] < 0
    ).sum(),

    "Water_Used_m3": (
        df_clean["Water_Used_m3"] < 0
    ).sum()
}

pd.Series(validity_checks, name="Invalid Values")
```

```python
# Check Farm_ID uniqueness

print("Total Records:", len(df_clean))
print("Unique Farm IDs:", df_clean["Farm_ID"].nunique())
```

```python
# Final missing value check

total_missing = df_clean.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
# Correlation between environmental conditions and yield

environmental_variables = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Soil_Moisture_pct"
]

environmental_correlation = (
    df_clean[
        environmental_variables + ["Yield_Tonnes_Ha"]
    ]
    .corr()["Yield_Tonnes_Ha"]
    .drop("Yield_Tonnes_Ha")
    .sort_values(ascending=False)
)

environmental_correlation
```

```python
# Correlation matrix of environmental conditions and yield

correlation_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

environmental_corr_matrix = df_clean[
    correlation_columns
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    environmental_corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Environmental Conditions and Agricultural Yield")
plt.tight_layout()
plt.show()
```
